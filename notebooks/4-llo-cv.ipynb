```python
from sklearn.datasets import make_classification
from sklearn.model_selection import cross_validate, LeaveOneOut
import warnings

warnings.filterwarnings("ignore")
```

```python
loo = LeaveOneOut()
```

```python
x, y = make_classification(n_samples=1000)
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(x)
```

```python
df["target"] = y
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
```

```python
rf = RandomForestClassifier()
logistic = LogisticRegression()
```

```python
rf_validator = cross_validate(
    estimator=rf,
    X=x,
    y=y,
    cv=5,
    scoring=["accuracy", "precision", "recall", "f1"],
    n_jobs=-1,
)
```

```python
lg_validator = cross_validate(
    estimator=logistic,
    X=x,
    y=y,
    cv=5,
    scoring=["accuracy", "precision", "recall", "f1"],
    n_jobs=-1,
)
```

```python
(
    lg_validator["test_accuracy"].mean(),
    lg_validator["test_precision"].mean(),
    lg_validator["test_recall"].mean(),
    lg_validator["test_f1"].mean(),
)
```

```python
(
    rf_validator["test_accuracy"].mean(),
    rf_validator["test_precision"].mean(),
    rf_validator["test_recall"].mean(),
    rf_validator["test_f1"].mean(),
)
```
