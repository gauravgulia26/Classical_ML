```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols("x")

f = sp.cos(x) ** 2 + sp.sin(x)

# Convert SymPy expression → NumPy function
f_numpy = sp.lambdify(x, f, "numpy")
```

```python
x_values = np.linspace(0, 20 * np.pi, 4096)

y_values = f_numpy(x_values)
```

```python
plt.plot(x_values, y_values)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("x + sin(x)")
plt.show()
```

```python
fft_values = np.fft.fft(y_values)
```

```python
magnitude = np.abs(fft_values)
```

```python
dx = x_values[1] - x_values[0]

frequencies = np.fft.fftfreq(len(x_values), d=dx)
```

```python
positive = frequencies >= 0

plt.plot(frequencies[positive], magnitude[positive])

plt.xlabel("Frequency")
plt.ylabel("Magnitude")
plt.title("FFT of x + sin(x)")
plt.xlim(0, 2)

plt.show()
```
