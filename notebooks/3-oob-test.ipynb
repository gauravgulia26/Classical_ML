```python
import numpy as np
import matplotlib.pyplot as plt

# Original dataset
n = 1000000

# Run 100 bootstrap simulations
unselected_percentages = []

for _ in range(100):
    sample = np.random.choice(n, size=n, replace=True)

    # Number of unique observations selected
    unique_selected = len(np.unique(sample))

    # Percentage not selected
    unselected = (n - unique_selected) / n * 100

    unselected_percentages.append(unselected)


# Plot distribution
plt.figure(figsize=(10, 5))

plt.hist(unselected_percentages, bins=10, edgecolor="black")

plt.axvline(36.8, linestyle="--", linewidth=4, label="Expected ≈ 36.8%", color="black")

plt.xlabel("Percentage of observations not selected")
plt.ylabel("Number of simulations")
plt.title("Bootstrap Sampling: Unselected Data Distribution")
plt.legend()

plt.show()
```
